```python
import pandas as pd

# 1. Cargar el CSV
df = pd.read_csv("fuel-prices-for-be-assessment.csv")

# 2. Inspección inicial
print("Primeras filas:")
print(df.head())

print("\nInformación general:")
print(df.info())

print("\nValores faltantes por columna:")
print(df.isnull().sum())

# 3. Estadísticas descriptivas
print("\nEstadísticas básicas:")
print(df.describe(include="all"))


# 6. Análisis rápido de cada columna
for col in df.columns:
    print(f"\nColumna: {col}")
    print("Tipo:", df[col].dtype)
    print("Valores únicos:", df[col].nunique())
    print("Ejemplos:", df[col].unique()[:5])

# 4. Distribución de valores de una columna categórica
print("\nDistribución de una columna categórica:")
print(df['nombre_columna'].value_counts())

# 5. Correlaciones entre variables numéricas
print("\nCorrelaciones:")
print(df.corr())

# 6. Análisis rápido de cada columna
for col in df.columns:
    print(f"\nColumna: {col}")
    print("Tipo:", df[col].dtype)
    print("Valores únicos:", df[col].nunique())
    print("Ejemplos:", df[col].unique()[:5])
```
